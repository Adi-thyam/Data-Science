```python
numbers = [10,11,12,13,14]
numbers.append(15)
print(numbers)
numbers.extend([16])
print(numbers)
numbers.insert(3,13)
print(numbers)
numbers.remove(13)
print(numbers)
numbers.pop()
print(numbers)
print(numbers.count(1))
print(numbers.index(14))
numbers.sort()
print(numbers)
numbers.reverse()
print(numbers)
numbers.clear()
print(numbers)
```

```python
marks = float(input('Enter the marks'))
if marks >= 90:
    print('Mark: ',marks,', Grade : A')
elif marks >=75:
    print('Mark: ',marks,', Grade : B')
elif marks >=50:
    print('Mark: ',marks,', Grade : C')
elif marks < 50:
    print('Mark: ',marks,', Grade : Fail')
else:
    print('Invalid number')
```

```python
number = int(input('Enter the number'))
if number > 0 and number %2 ==0:
    print(f'The number {number} is positive and Even')
else:
    print(f'The number {number} is negative or 0')
```
