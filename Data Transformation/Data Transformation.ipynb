```python
## We can add new features income_per_hour and age_group
df['income_per_hour'] = (df['capital_gain'] - df['capital_loss']) / (df['hours_per_week'] + 1)
df.head()
```

```python
def age_group(age):
    if age < 25:
        return 'Young'
    elif age < 45:
        return 'Adult'
    elif age < 65:
        return 'Middle_Age'
    else:
        return 'Senior'

df['age_group'] = df['age'].apply(age_group)
```
