```python
import scipy.stats as stats
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
# Given data
sample_mean = 3050
x_bar = 1000 + 5*600  
std_dev = 5 * 25
n = 25

# Z-statistic
z = (sample_mean - x_bar) / (std_dev / np.sqrt(n))

print("Z-statistic:", z)
```

```python
p_value = stats.norm.cdf(z)  
print("P-value:", p_value)
```

```python
critical_value = stats.norm.ppf(0.95)
print("Critical value:", critical_value)
```
