```python
import scipy.stats as stats
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
## Population standard deviation is unknown and sample size is <30 so go for t_test
## Given data
n = 15
data = [1.13,1.55,1.43,0.92,1.25,1.36,1.32,0.85,1.07,1.48,1.20,1.33,1.18,1.22,1.29]
x_bar = np.mean(data)
pop_std_dev = np.std(data)
confidence = 0.99
alpha = 1 - confidence

# t critical value
t_critical = stats.t.ppf(1 - alpha/2, df=n-1)

# margin of error
margin = t_critical * (pop_std_dev / np.sqrt(n))

# confidence interval
lower = x_bar - margin
upper = x_bar + margin

print("99% CI (t-distribution):", (lower, upper))
```

```python
## In here the population standard is known , so go for z_test
pop_std_dev = 0.2  # given population std

# z critical value
z_critical = stats.norm.ppf(1 - alpha/2)

# margin of error
margin_z = z_critical * (pop_std_dev / np.sqrt(n))

# confidence interval
lower_z = x_bar - margin_z
upper_z = x_bar + margin_z

print("99% CI (z-distribution):", (lower_z, upper_z))
```
