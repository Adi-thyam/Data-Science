```python
## GridSearchCv
```

```python
param_grid = {
    'neurons': [16, 32, 64],
    'activation': ['relu', 'tanh'],
    'learning_rate': [0.01, 0.001],
    'batch_size': [8, 16]
}
```

```python
def build_model(neurons=32, activation='relu', learning_rate=0.001):
    model4 = keras.Sequential([
        layers.Dense(neurons, activation=activation, input_shape=(60,)),
        layers.Dense(1, activation='sigmoid')
    ])

    model4.compile(
        optimizer=Adam(learning_rate=learning_rate),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model4
```

```python
results = []

for neurons in [16, 32, 64]:
    for activation in ['relu', 'tanh']:
        for lr in [0.01, 0.001]:

            model4 = build_model(neurons, activation, lr)

            history = model4.fit(
                x_train, y_train,
                epochs=20,
                batch_size=8,
                verbose=0
            )

            loss, acc = model4.evaluate(x_test, y_test, verbose=0)

            results.append({
                'neurons': neurons,
                'activation': activation,
                'learning_rate': lr,
                'accuracy': acc
            })
```

```python
y_pred4 = model4.predict(x_test)
y_pred4 = np.where(y_pred4>0.5,1,0)
y_pred4
```

```python
results_df = pd.DataFrame(results)
print(results_df.sort_values(by='accuracy', ascending=False))
```
